c) mergeListas: defina uma função que recebe duas listas de números ordenadas crescentemente como
argumento e devolve uma nova lista ordenada de forma crescente com os elementos das duas listas.

```python
#interativa 

def mergeListas(lista1,lista2):
    res = []
    for elem2 in lista2:
        for elem1 in lista1: 
            if elem1 < elem2:
                res.append(elem1)
            else:
                res.append(elem2)
    return res 
```

a) splitLista: defina uma função que recebe dois argumentos, uma lista de números e um número, e dá
como resultado um tuplo composto por duas listas, a primeira contendo os números menores ou iguais ao
número passado como argumento, e a segunda contendo os números maiores que o argumento.

```python
def splitLista(lista, n):
    res1 = []
    res2 = []
    for elem in lista: 
        if elem <= n:
            res1.append(elem)
        elif n < elem:
            res2.append(elem)
    return (res1,res2)

print(splitLista([1,2,3,4,5], 3))
```

b) multiplos23: defina uma função que calcula quantos múltiplos de 23 existem no intervalo de 1 a n em
que n é o argumento da função.

```python
def multiplos23(n):
    res = 0 
    for elem in range(1,n+1):
        if elem % 23 == 0:
            res = res + 1
    return res

print(multiplos23(100))
```

produto que for possível calcular multiplicando os 3 maiores inteiros da lista.

```python
def max3prod(lista):
    lista = sorted(lista)
    i = len(lista) - 1
    return lista[i]*lista[i-1]*lista[i-2]

print(max3prod([5,3,2,4,1]))
```
